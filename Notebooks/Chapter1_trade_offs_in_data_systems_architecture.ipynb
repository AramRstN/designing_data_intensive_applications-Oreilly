# Chapter 1: Trade-Offs in Data Systems Architecture
---

## Introduction: The philosophy of data systems

The central idea of this chapter is simple but extremely powerful:

>There is no perfect system design, only **trade-offs**.

Every decision in system design involves balancing competing factors:

- Performance vs cost
- Consistency vs availability
- Simplicity vs scalability

As systems grow, especially into **data-intensive systems**, these trade-offs become unavoidable. The goal of an engineer is not to eliminate trade-offs, but to **understand them deeply and choose wisely based on context**.

---

## What are data-intensive applications?

A **data-intensive application** is one in which the primary challenge is **handling data** rather than performing complex computations.

In contrast to compute-intensive systems (like simulations or scientific computing), data-intensive systems focus on:

- Storing large amounts of data
- Efficiently retrieving and updating data
- Managing concurrent access
- Handling failures and ensuring reliability

### Why this matters

At a small scale, a single machine is enough. But as users and data grow or requests increase, systems must become **distributed**, and complexity increases dramatically.

---

## Core building blocks of data systems

Modern applications are not built using a single system. Instead, they combine several specialized components:

**1. Databases**: These are the **core storage systems** where persistent data is stored and retrieved.

**2. Caches**: Caches store frequently accessed data in memory to reduce latency and avoid expensive recomputation.

**3. Search indexes**: They enable efficient querying beyond simple key lookups (e.g., full-text search, filtering).

**4. Stream processing systems**: These processes collect data in real time, reacting to events as they happen.

**5. Batch processing systems**: These processes periodically accumulate large volumes of data.

>💡 Key insight:
Real-world systems are compositions of multiple tools, not a single “perfect” system.

---

## Frontend vs Backend (Terminology)

Understanding this distinction is crucial for the rest of the book.

- **Frontend**: Runs on the user’s device (browser or mobile app)
- **Backend**: Runs on servers and handles data storage, business logic, and communication between services
      
### Important Observation

The frontend typically deals with one user at a time. The backend must:

- Handle millions of users
- Manage shared data
- Ensure consistency and performance

> This is why most data system complexity lies in the backend.

## Operational vs Analytical Systems

This is one of the most fundamental distinctions in data systems.

### Operational systems (OLTP)

Operational systems are responsible for **day-to-day operations**.

They:

- Handle user requests
- Store the current state of the system
- Support real-time interactions
Example:
- A user placing an order
- Updating a profile
- Logging into an app

These systems must be fast (low latency), reliable, and consistent.

### Analytical systems (OLAP)

Analytical systems are used for:

- Reporting
- Business Intelligence(BI)
- Machine learning

They do not serve end users directly. Instead, they help organizations **understand data and make decisions**.

Example queries:
- “What was total revenue last month?”
- “Which products are often bought together?”

## Data warehousing

To solve the conflict between OLTP and OLAP, organizations introduced **data warehouses**. A data warehouse is a system designed specifically for analytics, containing a **copy of data from operational systems**.

### ETL pipeline

The process of moving data into the warehouse consists of:

**1. Extract**: Data is taken from operational systems.

**2. Transform**: Data is cleaned, structured, and reshaped.

**3. Load**: Data is stored in the warehouse.

This pipeline ensures that:

- Analytical queries do not affect operational systems
- Data is optimized for analysis

### Why not query OLTP directly?

Because:

- Data is spread across systems (data silos)
- Queries are expensive
- Performance would degrade
- Security concerns exist

## From data warehouse to data lake

As data science evolved, data warehouses became insufficient.

A **data lake** is a centralized repository that stores raw data in its original form.

Unlike data warehouses:

- No strict schema is required
- Supports all data types (text, images, logs, etc.)

### Why data lakes?

Data scientists often need raw data, flexibility, and custom transformations. For example:

- Feature engineering for ML
- NLP on text data
- Image processing

These are difficult in traditional SQL-based warehouses.

#### Schema-on-read/write
- Warehouse: schema-on-write
- Data lake: schema-on-read

Meaning: Structure is applied when reading, not when storing.

## Systems of record vs Derived data

This concept explains data flow in systems.

### System of record

This is the **source of truth**.

- Data is written here first
- It is authoritative
- If conflicts exist → this is correct

### Derived data

Derived data is created by:

- Transforming
- Aggregating
- Indexing

Examples:

- Cache
- Search index
- Materialized views
- ML models

>💡 Key insight
Derived data improves performance but introduces redundancy.
However, it can always be recomputed, and it is not the source of truth.

## Cloud vs Self-hosting
### The core question: Build or buy?

At a high level, choosing between cloud and self-hosting is not just technical. It is a **business decision**:

>Should you **build and operate systems yourself**, or **outsource them to a provider**?

This decision affects:

- Cost structure (CAPEX vs OPEX)
- Team skills required
- Speed of development
- Control over infrastructure

### The spectrum of options

Rather than a binary choice, there is a **continuum**:

**1. Fully self-hosted (on-premises)**: You own hardware and operate everything.

**2. Self-hosted on cloud infrastructure (IaaS)**: You rent VMs but manage software yourself.

**3. Managed services (PaaS)**: Provider manages infrastructure and software.

**4. SaaS (fully managed)**: You only use an API or UI.

### Advantages of cloud services
**1. Reduced operational burden**: Running distributed systems is hard:

- Deployment
- Monitoring
- Scaling
- Failure recovery

Cloud providers specialize in this and amortize expertise across customers.

**2. Elastic scalability**: Cloud allows scaling up during peak load and scaling down during idle time. This is especially useful for:

- Analytical workloads
- Spiky traffic systems

> You avoid paying for idle infrastructure.

**3. Faster time-to-market**: Instead of hiring ops engineers and setting up infrastructure, you can:

- Use managed databases
- Deploy quickly

**4. Access to advanced infrastructure**: Cloud gives easy access to distributed storage, global replication, and GPUs / specialized hardware.

### Disadvantages of cloud services

**1. Loss of control**: You cannot modify internal behavior or tune low-level performance deeply. If the system doesn’t support your use case, you are stuck.

**2. Vendor lock-in**: Many cloud services use proprietary APIs and lack standardization. Switching providers becomes costly and complex.

**3. Debugging limitations**: In self-hosting, you have logs, metrics, and system-level visibility. In the cloud, there is limited observability, and it's a black-box system.

**4. Cost uncertainty**: Cloud pricing is Pay-as-you-go, and it is hard to predict at scale. Thus, cost optimization becomes a continuous engineering task.

**5. External risks**: Some risks are out of your control in case of service outages, pricing changes. There are also possible political/geographical risks.

>💡 Key insight:
Cloud is not always cheaper—it is often more flexible, not necessarily more cost-efficient.

Cloud-native systems are not just running in the cloud. They are designed from the ground up to leverage cloud properties.

### Core design principles
**1. Separation of storage and compute**

In traditional systems, data is stored on a local disk, and computation happens on the same machine. (Tight coupling)

In cloud-Native Systems, storage is external (e.g., object storage), and computations are independent workers. (Loose coupling)

This is important because it enables independent scaling, fault isolation, and better resource utilization. The trade-off would be that the data must travel over the network, resulting in increased latency.

**2. Object storage as a foundation**

Cloud systems often rely on S3-like storage. The characteristics are:

- Highly durable
- Distributed automatically
- Simple API (read/write objects)
>💡 Insight:
Object storage trades flexibility (filesystem features) for **scalability and reliability**.

**3. Multitenancy**

It means multiple customers share the same infrastructure.The benefits are:
- Higher utilization
- Lower cost per user

The challenges are:
- Performance isolation
- Security isolation
- Fair resource allocation
  
**4. Layered architecture**

Cloud systems are built in layers:

- Storage layer (S3)
- Compute layer (Spark, query engines)
- Application layer
  
>💡 Insight:
Cloud encourages building systems as composable services.

## Operations in the cloud era
### Evolution of operations

The traditional roles, like Sysadmins and DBAs replaced by modern roles like DevOps and SRE (Site Reliability Engineering).

### Shift in responsibilities
Before (Self-hosting), the focus was on machines, hardware provisioning, and disk management. Now (Cloud), the focus shifts to service integration, monitoring system behavior, and cost optimization

### New operational principles

**1. Automation first** is about avoiding manual operations and using repeatable processes.

**2. Ephemeral infrastructure** means machines are temporary and should be replaced instead of repaired.

**3. Observability** is about the requirement of visibility into metrics, logs, and traces because systems are distributed and dynamic.

**4. Cost awareness** says the capacity planning becomes the financial planning

>💡 Key insight:
Cloud reduces infrastructure work, but **does not eliminate operations. It changes it**.

## Distributed vs Single-node systems
### What is a distributed system?

A system where multiple machines (nodes) communicate over a network.

### Why use distributed systems?

**1. Scalability**: When the data is more than the capacity of a single machine, or the required computation is more than the capacity of a single CPU.

**2. Fault tolerance**: Redundancy allows the system to continue even if nodes fail.

**3. Geographic distribution**: It can help reduce latency and serve global users.

**4. Elasticity**: It is feasible to scale resources dynamically

### Fundamental challenges
**1. Network unreliability**: Unlike local calls:

- Network calls can fail
- Timeouts occur
- Messages may be lost
  
**2. Latency**: A network is orders of magnitude slower than memory.

**3. Partial failures**: If some nodes fail, others don’t, the system enters inconsistent states.

**4. Debugging complexity**: Hard to answer where the request failed?
 Or which service caused the delay?
 
**5. Data consistency**: With multiple nodes, there are multiple copies of data. Keeping them consistent is difficult.

>💡 Critical insight:
Distributed systems add complexity non-linearly.

>Rule of thumb: If a system can work on a single machine, prefer that.

## Microservices architecture

Microservices are primarily about organizational scalability, not just technical scalability.

### Structure

The system is split into independent services, each with a single responsibility, communicating via APIs

### Benefits
**1. Independent development**: Teams can work autonomously and deploy independently.

**2. Technology flexibility**: Each service can use different language and different databases.

**3. Scalability per service**: Scale only what needs scaling.

### Challenges
**1. Increased complexity**
- Network communication everywhere
- More moving parts

**2. Testing difficulty**: Need to simulate multiple services.

**3. API evolution problems**: Changing APIs can break dependent services.

**4. Data consistency issues**: Each service owns its database. Cross-service consistency becomes difficult.

>💡 Key insight:
Microservices trade simplicity for scalability of teams and systems.

---

## Serverless (Function-as-a-Service)
### Concept

You write the small functions, Cloud handles execution, scaling, and infrastructure.

### Advantages
**1. No infrastructure management**: No need to provision servers and manage scaling.

**2. Cost efficiency (in some cases)**: You pay only for execution time.

**3. Automatic scaling**: Handles sudden spikes.

### Limitations
**1. Execution constraints**
- Time limits
- Resource limits
  
**2. Cold start latency**
The first request is slower.

**3. Less control**
Limited customization

>💡 Insight:
Serverless is ideal for event-driven workloads, not all systems.

## Cloud vs Supercomputing
### Fundamental difference
| Cloud Systems              | Supercomputers      |
| -------------------------- | ------------------- |
| Serve users continuously   | Run batch jobs      |
| Handle failures gracefully | Restart on failure  |
| Distributed globally       | Usually centralized |

>💡 Insight:
Cloud prioritizes: availability and flexibility. HPC prioritizes:raw performance.

## Data Systems, Law, and Society
### Why this matters?

Data systems are not just technical. They affect privacy, society, and ethics.

### Regulations
**GDPR (EU)**: Gives users rights to access, delete, or control data.

**Key principle:** Right to Be Forgotten

>This creates a technical challenge: How do you delete data in systems designed to be immutable?

**Key principle:** Data Minimization

>Only collect and store necessary data

### Hidden costs of data

Storing data has risks:

- Legal liability
- Security breaches
- Ethical consequences
>💡 Insight:
“More data is always better” is not always true.

### Engineering responsibility

Engineers must consider societal impact, ensure compliance, design responsibly.

---

## Final Summary
### Core lessons
**1. Everything is a trade-off**. No design is universally best.

**2. Separation is key**
- OLTP vs OLAP
- Storage vs compute
- Services vs responsibilities
  
**3. Complexity comes with scale**. Distributed systems, microservices, and cloud systems all increase complexity.

**4. Simplicity is underrated**. Prefer simple systems unless complexity is justified.

**5. Data has responsibility**. Beyond engineering, there are also legal aspects, ethical aspects, and societal implications.
