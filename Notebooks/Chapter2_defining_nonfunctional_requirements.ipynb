# Chapter 2: Defining Nonfunctional Requirements

---

## The Big Picture of This Chapter

In software engineering, people often focus primarily on functional requirements:

- What should the application do?
- What features should it provide?
- What buttons and APIs should exist?

**But** in real systems, functionality alone is not enough.

A system can technically work, produce correct results, and still be considered a failure. Why? Because users also care about: speed, reliability, scalability, uptime, maintainability, and operational stability.

These are called **nonfunctional requirements**, and this chapter explains how to reason about them systematically.

The chapter introduces four major system qualities:

- Performance: Is the system fast enough?
- Reliability: Does it keep working when things go wrong?
- Scalability: Can it handle growth? 
- Maintainability: Can humans continue evolving it?
